```python
import os
from data_preparation import load_data

df = load_data("Sarcasm_Headlines_Dataset_v2.json",columns=["headline", "is_sarcastic"])

print(df.head())
```

```python
#check for missing values
missing_values = df.isnull().sum()
print("Missing values in each column:")
print(f"headline: {missing_values['headline']}")
print(f"is_sarcastic: {missing_values['is_sarcastic']}\n")

#check for duplicates
duplicates = df["headline"].duplicated().sum()
print(f"Number of duplicate headlines: {duplicates}")

#remove duplicates and missing values
df = df.dropna(subset=["headline", "is_sarcastic"])
df = df.drop_duplicates(subset="headline")
df = df.reset_index(drop=True)
```
